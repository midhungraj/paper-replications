```python
import torch
import torch.nn as nn

torch.manual_seed(42)

# -------------------------
# Input
# -------------------------

tokens = ["the", "cat"]

# Fake token embeddings
X = torch.tensor([
    [1.0, 0.0, 1.0, 0.0],   # the
    [0.0, 1.0, 0.0, 1.0]    # cat
])

d_model = 4
d_k = 4

# -------------------------
# Learned projections
# -------------------------

W_Q = nn.Linear(d_model, d_k, bias=False)
W_K = nn.Linear(d_model, d_k, bias=False)
W_V = nn.Linear(d_model, d_model, bias=False)

# -------------------------
# Q K V
# -------------------------

Q = W_Q(X)
K = W_K(X)
V = W_V(X)

print("\nQ:")
print(Q)

print("\nK:")
print(K)

print("\nV:")
print(V)

# -------------------------
# Attention scores
# -------------------------

scores = Q @ K.T

print("\nRaw scores:")
print(scores)

# -------------------------
# Scaling
# -------------------------

scores = scores / (d_k ** 0.5)

print("\nScaled scores:")
print(scores)

# -------------------------
# Attention weights
# -------------------------

attention = torch.softmax(
    scores,
    dim=-1
)

print("\nAttention weights:")
print(attention)

# -------------------------
# Contextual representations
# -------------------------

output = attention @ V

print("\nAttention output:")
print(output)
```
