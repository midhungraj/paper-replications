```python
import math
import torch
```

```python
position = torch.arange(5)
print(position), print(position.size())
```

```python
position = torch.arange(5).float().unsqueeze(1)
print(position), print(position.size())
```

```python
d_model = 4
```

```python
pe = torch.zeros(5, 4)
pe
```

```python
div_term = torch.exp(torch.arange(0, d_model, 2).float() 
* (-math.log(10000.0) / d_model))

div_term
```

```python
pe[:, 0::2] = torch.sin(position * div_term)
pe[:, 1::2] = torch.cos(position * div_term)
```

```python
pe[:, 0::2]
```

```python
pe[:, 1::2]
```
